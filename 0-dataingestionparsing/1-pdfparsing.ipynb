```python
## Here's full program that use the document file as pdf format and return the chunks
```

```python
from typing import List
from langchain_core.documents import Document
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

class PDF:

    def __init__(self, chunk_size=1000, chunk_overlap=100):

        # Validate BEFORE use
        if chunk_size <= 0:
            raise ValueError(f"chunk_size must be > 0, got {chunk_size}")
        if chunk_overlap < 0:
            raise ValueError(f"chunk_overlap must be >= 0, got {chunk_overlap}")

        self.chunk_size = chunk_size
        self.chunk_overlap = chunk_overlap

        self.text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=self.chunk_size,
            chunk_overlap=self.chunk_overlap,
            separators=["\n\n", "\n", " ", ""]
        )

    def pdf_preprocessing(self, pdf_path) -> List[Document]:
        try:
            pdf_loader = PyPDFLoader(pdf_path)
            pdf_content = pdf_loader.load()

            pdf_preprocessing_chunks = []

            for page_num, page in enumerate(pdf_content):
                chunks = self.text_splitter.create_documents(
                    texts=[page.page_content],
                    metadatas=[{
                        **page.metadata,
                        "page_number": page_num + 1
                    }]
                )
                pdf_preprocessing_chunks.extend(chunks)

            print(f"Generated {len(pdf_preprocessing_chunks)} chunks")
            return pdf_preprocessing_chunks

        except Exception as error:
            print(f"Error in PDF preprocessing: {error}")
            return []  # Safe fallback
```

```python
pdf_object = PDF()
result = pdf_object.pdf_preprocessing("../documents/invoice.pdf")

print(len(result))
print(result[1].page_content)
```
